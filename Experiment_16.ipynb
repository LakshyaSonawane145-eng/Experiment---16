```python
#1.Tokenization
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

```

```python
from nltk.tokenize import word_tokenize

text = "Natural language processing is interesting"

tokens = word_tokenize(text)

print(tokens)
```

```python
#Sentence Tokenization
from nltk.tokenize import sent_tokenize

text = "Python is easy. It is widely used in data science."

sentences = sent_tokenize(text)

print(sentences)
```

```python
#Stop Word Removal
#Stop words such as is , the , and , in are removed

from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

tokens = word_tokenize(text)

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print(filtered_tokens)
print(stop_words)
```

```python
#Stemming
#Stemming reduces words to their root form

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["coding","watched","gone","studies","playing","running", "runs", "ran"]

for w in words:
  print(stemmer.stem(w))
```

```python
#Part-of-Speech (POS) Tagging
#POS tagging identifies grammatical roles of words.

import nltk
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag

word = word_tokenize("Python is powerful programming language")
pos_tag1 = pos_tag(word)
print(pos_tag1)
```

```python
#Word Frequency Count
from nltk.probability import FreqDist

words = word_tokenize(text)

fd = FreqDist(words)
print(fd)
print(fd.most_common())
```

```python
#Lemmatization
#Lemmatization converts words to their dictionary form

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["coding","watched","gone","studies","playing","running", "runs", "ran"]

for w in words:
  print(lemmatizer.lemmatize(w))
```
